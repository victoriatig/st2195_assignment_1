## Printing integers from 1 to 10
The folowing Python code uses a for loop to print all integers from 1 to 10, including both 1 and 10.




```python
for i in range(1,11):
    print(i)
```


The for loop starts at 1 and repeats until it reaches 10.

The range(1,11) function generates integers from 1 to 10.
